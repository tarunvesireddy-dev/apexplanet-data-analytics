```python
"""
db_utils.py
Reusable utility functions for connecting to and querying the Superstore SQLite database.
"""

from sqlalchemy import create_engine
import pandas as pd

DB_PATH = "../data/superstore.db"

def get_engine(db_path=DB_PATH):
    """Create and return a SQLAlchemy engine connected to the database."""
    return create_engine(f'sqlite:///{db_path}')

def run_query(query, engine=None):
    """Run a SQL query and return the result as a DataFrame."""
    if engine is None:
        engine = get_engine()
    return pd.read_sql(query, engine)

def top_n_by_column(table, column_to_rank, n=5, ascending=False, engine=None):
    """Return the top N rows from a table, ranked by a given column."""
    order = "ASC" if ascending else "DESC"
    query = f"SELECT * FROM {table} ORDER BY {column_to_rank} {order} LIMIT {n}"
    return run_query(query, engine)

def group_summary(table, group_col, agg_col, agg_func="SUM", engine=None):
    """Return a grouped summary (e.g. SUM/AVG/COUNT of agg_col grouped by group_col)."""
    query = f"""
    SELECT {group_col}, {agg_func}({agg_col}) as {agg_func}_{agg_col}
    FROM {table}
    GROUP BY {group_col}
    ORDER BY {agg_func}_{agg_col} DESC
    """
    return run_query(query, engine)
```
