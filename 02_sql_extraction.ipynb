```python
import sqlite3
import pandas as pd

# Load your cleaned data
df = pd.read_csv('../data/superstore_cleaned.csv')

# Create/connect to a SQLite database file
conn = sqlite3.connect('../data/superstore.db')

# Write the dataframe into a table called 'sales'
df.to_sql('sales', conn, if_exists='replace', index=False)

print("Data imported successfully!")
```

```python
pd.read_sql("SELECT * FROM sales LIMIT 5", conn)
```

```python
df.columns = df.columns.str.replace(' ', '_').str.replace('-', '_')
df.to_sql('sales', conn, if_exists='replace', index=False)
```

```python
pd.read_sql("SELECT * FROM sales LIMIT 3", conn)
```

```python
# SELECT + WHERE
pd.read_sql("SELECT * FROM sales WHERE Category='Furniture' LIMIT 5", conn)
```

```python
# ORDER BY
pd.read_sql("SELECT Product_Name, Sales FROM sales ORDER BY Sales DESC LIMIT 10", conn)
```

```python
# LIMIT + GROUP BY
pd.read_sql("SELECT Region, SUM(Sales) as Total_Sales FROM sales GROUP BY Region ORDER BY Total_Sales DESC", conn)
```

```python
# GROUP BY + HAVING
pd.read_sql("""
SELECT Sub_Category, SUM(Profit) as Total_Profit 
FROM sales 
GROUP BY Sub_Category 
HAVING Total_Profit < 0
""", conn)
```

```python
# JOIN practice — since Superstore is a single flat table, you won't have a natural join here.
# We can simulate one by creating a small "regions" lookup table to practice JOIN syntax:
regions_df = pd.DataFrame({
    'Region': ['East', 'West', 'Central', 'South'],
    'Manager': ['Alice', 'Bob', 'Carol', 'Dave']
})
regions_df.to_sql('regions', conn, if_exists='replace', index=False)

pd.read_sql("""
SELECT s.Region, s.Sales, r.Manager
FROM sales s
JOIN regions r ON s.Region = r.Region
LIMIT 5
""", conn)
```

```python
pd.read_sql("""
SELECT Order_ID, Product_Name, Sales
FROM sales
WHERE Sales > (SELECT AVG(Sales) FROM sales)
ORDER BY Sales DESC
LIMIT 10
""", conn)
```

```python
pd.read_sql("""
WITH category_profit AS (
    SELECT Category, SUM(Profit) as Total_Profit
    FROM sales
    GROUP BY Category
)
SELECT * FROM category_profit
ORDER BY Total_Profit DESC
""", conn)
```

```python
pd.read_sql("""
SELECT Category, Product_Name, Sales,
       RANK() OVER (PARTITION BY Category ORDER BY Sales DESC) as Sales_Rank
FROM sales
LIMIT 20
""", conn)
```

```python
pd.read_sql("""
WITH ranked AS (
    SELECT Category, Product_Name, Sales,
           ROW_NUMBER() OVER (PARTITION BY Category ORDER BY Sales DESC) as rn
    FROM sales
)
SELECT Category, Product_Name, Sales
FROM ranked
WHERE rn = 1
""", conn)
```

```python
pd.read_sql("""
SELECT Order_Date, Sales,
       LAG(Sales) OVER (ORDER BY Order_Date) as Previous_Sales
FROM sales
LIMIT 10
""", conn)
```

```python
conn.execute("""
CREATE VIEW IF NOT EXISTS category_profit_summary AS
SELECT Category, SUM(Sales) as Total_Sales, SUM(Profit) as Total_Profit
FROM sales
GROUP BY Category
""")

pd.read_sql("SELECT * FROM category_profit_summary", conn)
```

```python
from sqlalchemy import create_engine

engine = create_engine('sqlite:///../data/superstore.db')

# Test it
pd.read_sql("SELECT * FROM sales LIMIT 3", engine)
```

```python
# Q1: Top 5 products by total sales
pd.read_sql("""
SELECT Product_Name, SUM(Sales) as Total_Sales
FROM sales GROUP BY Product_Name
ORDER BY Total_Sales DESC LIMIT 5
""", engine)
```

```python
# Q2: Monthly sales trend
pd.read_sql("""
SELECT strftime('%Y-%m', Order_Date) as Month, SUM(Sales) as Total_Sales
FROM sales GROUP BY Month ORDER BY Month
""", engine)
```

```python
# Q3: Customer segmentation by total spend
pd.read_sql("""
SELECT Customer_Name, SUM(Sales) as Total_Spend
FROM sales GROUP BY Customer_Name
ORDER BY Total_Spend DESC LIMIT 10
""", engine)
```

```python
# Q4: Most profitable region
pd.read_sql("""
SELECT Region, SUM(Profit) as Total_Profit
FROM sales GROUP BY Region ORDER BY Total_Profit DESC
""", engine)
```

```python
# Q5: Average discount by category
pd.read_sql("""
SELECT Category, AVG(Discount) as Avg_Discount
FROM sales GROUP BY Category
""", engine)
```

```python
# Q6: Which ship mode is used most often
pd.read_sql("""
SELECT Ship_Mode, COUNT(*) as Order_Count
FROM sales GROUP BY Ship_Mode ORDER BY Order_Count DESC
""", engine)
```

```python
# Q7: Top 5 customers by number of orders (not just spend)
pd.read_sql("""
SELECT Customer_Name, COUNT(DISTINCT Order_ID) as Num_Orders
FROM sales GROUP BY Customer_Name
ORDER BY Num_Orders DESC LIMIT 5
""", engine)
```

```python
# Q8: Loss-making orders (profit < 0) by category
pd.read_sql("""
SELECT Category, COUNT(*) as Loss_Orders
FROM sales WHERE Profit < 0
GROUP BY Category ORDER BY Loss_Orders DESC
""", engine)
```

```python
# Q9: Average shipping delay (days between Order_Date and Ship_Date) by Ship_Mode
pd.read_sql("""
SELECT Ship_Mode, AVG(julianday(Ship_Date) - julianday(Order_Date)) as Avg_Delay_Days
FROM sales GROUP BY Ship_Mode
""", engine)
```

```python
# Q10: Top-selling sub-category per region (window function)
pd.read_sql("""
WITH ranked AS (
    SELECT Region, Sub_Category, SUM(Sales) as Total_Sales,
           RANK() OVER (PARTITION BY Region ORDER BY SUM(Sales) DESC) as rnk
    FROM sales
    GROUP BY Region, Sub_Category
)
SELECT Region, Sub_Category, Total_Sales
FROM ranked WHERE rnk = 1
""", engine)
```
